# Binance Advanced ETL

### Introduction

The purpose of this project is to design and implement a realistic ETL pipeline for financial market data using a public cryptocurrency API. The pipeline ingests market data from Binance, processes and normalizes it, and stores it in a PostgreSQL data warehouse designed for analytical workloads.


### Data Model Design

* Dimension tables:
    * assets: stores unique cryptocurrencies
    * markets: represents trading pairs and exchange metadata

* Fact tables:
    * candles: time-series OHLCV data
    * trades: aggregated trade records
    * tickers_24h: daily market snapshots

### Files
- scraper.py -> downloads API data and writes processed jsons
- clean-data.py -> cleans data, ahndles duplicates, nulls...
- insert_to_db.py -> loads into PostgreSQL (idempotent)
- schema.sql -> creates schema, tables, partitions
- queries.sql -> queries

### Configure
```
{
  "base_url": "https://api.binance.com",  // Binance Spot REST API
  "quote_asset": "USDT",                  // Filters markets (USDT pairs only)
  "top_n_markets": 20,                    // Selects top markets by 24h volume
  "kline_interval": "5m",                 // 5m → 288 rows/day
  "start_date_utc": "2025-05-01",
  "end_date_utc": null,
  "trades_per_market": 5000                // Trade volume cap per market
}
```

## Run order
1) schema.sql (DataGrip)
2) scraper.py (PyCharm)
3) clean_data.py (PyCharm)
4) insert_to_db.py (PyCharm)
5) queries.sql (DataGrip)

### Analytics and SQL Queries

* JOIN operations across dimensions and facts
* Aggregations and grouping
* Advanced filtering using HAVING and BETWEEN
* Window functions for ranking and running totals
* Conditional logic with CASE
* NULL handling using COALESCE
* Performance analysis using EXPLAIN ANALYZE
